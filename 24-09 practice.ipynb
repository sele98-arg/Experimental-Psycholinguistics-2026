```python
name = "Sele"
birth = "1998"
```

```python
def make_username(first, year):
    username = first.lower().replace("","") + year
    print(username)
```

```python
make_username(name, birth)
```

```python
def make_username(int1, int2):
    username = int1.lower().replace("", "") + int2
    print(username)
```

```python
make_username(name, birth)
```
